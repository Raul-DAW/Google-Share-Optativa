Funciones principales del programa💗:


cargarModelo():
Carga los archivos del modelo entrenado (`model.json` y `metadata.json`) usando Teachable Machine. Comprueba primero si el modelo ya está cargado para no cargarlo varias veces.


iniciarCamara():
Activa la cámara del dispositivo. Crea la webcam, pide permiso al usuario, empieza a mostrar la imagen y llama a la función que analiza la cámara.


comprobarCamara():
Analiza continuamente la imagen de la cámara. Actualiza cada fotograma, envía la imagen al modelo y obtiene las predicciones.


pararCamara():
Detiene la cámara y deja de analizar imágenes. También borra la imagen de la cámara de la página.


cambiarPorcentaje():
Lee el valor seleccionado en la barra de porcentaje y lo muestra en pantalla para que el usuario vea el mínimo elegido.


mostrarResultado():
Busca cuál de las predicciones tiene el porcentaje más alto. Después compara ese porcentaje con el mínimo seleccionado y muestra si el objeto se considera detectado o no.


subirImagen():
Permite seleccionar una imagen del ordenador. Carga esa imagen, la envía al modelo y muestra el resultado de la predicción.


guardarHistorial():
Busca el resultado con mayor probabilidad y guarda su nombre, porcentaje, origen y fecha usando `localStorage`.


mostrarHistorial():
Lee los resultados guardados en `localStorage` y los muestra en la página uno debajo de otro.


borrarHistorial():
Elimina los resultados guardados en `localStorage` y actualiza la zona del historial para dejarla vacía.
